```python
# list creation
# manual
l1 = [1,2,3,4,5,6]

# automatic / comprehension
l2 = [x for x in range(1, 7)]
print(l1, l2, sep=", ")
```

```python
length = len(l1)
print(f"Length of list is: {length}")
print(f"Indexes present of list is: {length - 1}")

# adding element at the end of the list
# l1.append(7)
# l2.append(7)
print(l1, l2, sep=", ")

# removing last element of list
l1.pop()
l2.pop()
print(l1, l2, sep=", ")
```

```python
# updating list element
l1[2] = 98
l2[2] = 100
print(l1, l2, sep=", ")
```

```python
# reversing list
# l1.reverse()
# l2.reverse()
# print(l1, l2, sep=", ")

# sorting
l1.sort()
l2.sort()
print(l1, l2, sep=", ")

# slicing
# l1 = l1[:]
# l2 = l2[:]
print(l1[1:-1], l2[1:-1], sep=", ")
```
