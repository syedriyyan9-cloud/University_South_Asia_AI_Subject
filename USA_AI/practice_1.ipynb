### Arithmetic Operators

```python
n1 = 10
n2 = 4
print(f"Addition gives: {n1 + n2}")
print(f"Subtraction gives: {n1 - n2}")
print(f"Multiplication gives: {n1 * n2}")
print(f"Division gives: {n1 / n2}") # gives float
print(f"Int Division gives: {n1 // n2}") # gives int
print(f"Exponentional gives: {n1 ** n2}") # Exponentional
print(f"Remainder is: {n1 % n2}") # Exponentional
```

### Conditional Logic
- 3 types of conditional statements

```python
even_or_odd = 7
if even_or_odd % 2 == 0:
    print("EVEN")
else:
    print("ODD")

year = 2017
if year % 4 == 0:
    print("Leap year")
else:
    print("Not leap year")
```

```python
marks = 82
if marks > 90:
    print("You Got: A")
elif marks > 80:
    print("You Got: B")
elif marks > 70:
    print("You Got: C")
elif marks > 60:
    print("You Got: D")
else:
    print("You Got: E")
```
