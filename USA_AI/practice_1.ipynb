### Arithmetic Operators

```python
n1 = 10
n2 = 4
print(f"Addition gives: {n1 + n2}")
print(f"Subtraction gives: {n1 - n2}")
print(f"Multiplication gives: {n1 * n2}")
print(f"Division gives: {n1 / n2}") # gives float
print(f"Int Division gives: {n1 // n2}") # gives int
print(f"Exponentional gives: {n1 ** n2}") # Exponentional
print(f"Remainder is: {n1 % n2}") # Exponentional
```
