### **Instructions to Install Anaconda or Miniconda for Jupyter Notebooks on Mac or Windows**

#### **Option 1: Installing Anaconda**

Anaconda is a complete package that comes with Python, Jupyter Notebook, and many other useful libraries for data science and machine learning.

1. **Download Anaconda:**
   - Visit the [Anaconda distribution page](https://www.anaconda.com/products/distribution).
   - Choose the version appropriate for your operating system (Windows or macOS).
   - Click the download button to start the download.

2. **Install Anaconda:**
   - **Windows:**
     - Double-click the downloaded `.exe` file.
     - Follow the instructions in the setup wizard.
     - It's recommended to check the option to "Add Anaconda to my PATH environment variable" and "Register Anaconda as my default Python".
   - **Mac:**
     - Open the downloaded `.pkg` file.
     - Follow the installation instructions.

3. **Verify Installation:**
   - Open a terminal or command prompt.
   - Type `conda --version` to verify the installation. You should see the version number of Anaconda installed.
   - Type `jupyter notebook` to start Jupyter Notebook.

#### **Option 2: Installing Miniconda**

Miniconda is a lightweight version of Anaconda that includes only the conda package manager and Python. You can install additional packages as needed.

1. **Download Miniconda:**
   - Visit the [Miniconda installation page](https://docs.conda.io/en/latest/miniconda.html).
   - Choose the installer appropriate for your operating system (Windows or macOS).
   - Click the download link for your system.

2. **Install Miniconda:**
   - **Windows:**
     - Double-click the downloaded `.exe` file.
     - Follow the setup wizard instructions.
     - It's recommended to check the option to "Add Miniconda to my PATH environment variable".
   - **Mac:**
     - Open the downloaded `.sh` file via the terminal:
       ```bash
       bash Miniconda3-latest-MacOSX-x86_64.sh
       ```
     - Follow the on-screen instructions.

3. **Set Up Jupyter Notebook:**
   - Open a terminal or command prompt.
   - Create a new environment (optional but recommended):
     ```bash
     conda create -n myenv python=3.9
     conda activate myenv
     ```
   - Install Jupyter Notebook:
     ```bash
     conda install jupyter
     ```
   - Start Jupyter Notebook:
     ```bash
     jupyter notebook
     ```

### Order of operations

The arithmetic we learned in primary school has conventions about the order in which operations are evaluated. Some remember these by a mnemonic such as **PEMDAS** - **P**arentheses, **E**xponents, **M**ultiplication/**D**ivision, **A**ddition/**S**ubtraction.

Python follows similar rules about which calculations to perform first. They're mostly pretty intuitive.

`````python
8 - 3 + 2
`````

`````python
-3 + 4 * 2
`````

Sometimes the default order of operations isn't what we want:

`````python
hat_height_cm = 25
my_height_cm = 190
# How tall am I, in meters, when wearing my hat?
total_height_meters = hat_height_cm + my_height_cm / 100
print("Height in meters =", total_height_meters, "?")
`````

Parentheses are useful here. You can add them to force Python to evaluate sub-expressions in whatever order you want.

`````python
total_height_meters = (hat_height_cm + my_height_cm) / 100
print("Height in meters =", total_height_meters)
`````

##Logical Operators

The AND operator is a logical operator used to combine two Boolean expressions. It evaluates both expressions and returns True only if both expressions are True. If either of the expressions is False, the result of the and operation is False.

`````python
x = 5
y = 10

result = (x > 3) and (y < 20)
print(result)
`````

Here, both (x > 3) and (y < 20) are True, so the result of the and operation is True.

`````python
x = 5
y = 25

result = (x > 3) and (y < 20)
print(result)
`````

In this case, while (x > 3) is True, (y < 20) is False, so the overall result is False.

`````python
age = 25
has_ticket = True

can_enter = (age >= 18) and has_ticket
print(can_enter)
`````

the OR operator is another logical operator used to combine Boolean expressions. It evaluates two expressions and returns True if at least one of the expressions is True. If both expressions are False, then the result of the or operation is False.

`````python
a = True
b = False

result = a or b
print(result)
`````

### Set

In Python, a set is a built-in data type that represents an unordered collection of unique elements. Unlike lists and tuples, sets do not allow duplicate values and do not maintain any order of the elements. Sets are mutable, meaning you can add or remove items, but the order of items is not preserved.

Key Characteristics of Sets:


*   Unordered: Sets do not maintain the order of elements.

*   Unique Elements: Each element in a set must be unique.
*   Mutable: You can add or remove elements from a set.


*   No Indexing: Sets do not support indexing or slicing.







`````python
my_set = {1, 2, 3, 4}
`````

You can create a set by placing comma-separated values inside curly braces {} or by using the set() constructor.

`````python
empty_set = set()
print(empty_set)
`````

`````python
numbers = {1, 2, 3}
numbers.add(4)
print(numbers)
`````

`````python
numbers = {1, 2, 3, 4}
numbers.remove(2)
print(numbers)
`````

`````python
numbers.discard(5)  # This will not raise an error even if 5 is not in the set.
`````

Note: If the element to be removed is not present, remove() will raise a KeyError. To avoid this, use discard():

`````python
numbers = {1, 2, 3, 4}
element = numbers.pop()
print(element)  # Output: An arbitrary element (since sets are unordered)
print(numbers)  # Output: The set with one element removed
`````

`````python
my_list = [1, 2, 2, 3, 4, 4, 5]
unique_elements = set(my_list)
print(unique_elements)
`````

### Tuples

In Python, a tuple is a built-in data type that represents an ordered collection of elements. Unlike lists, tuples are immutable, which means that once a tuple is created, its contents cannot be changed. Tuples can contain elements of different data types and can be nested within other tuples.

Key Characteristics of Tuples:


*   Ordered: Tuples maintain the order of elements.
*   Immutable: Once created, the elements of a tuple cannot be changed, added, or removed.

*   Indexed: Elements in a tuple can be accessed using indices, starting from 0.
*   Can Contain Different Data Types: Tuples can store elements of different types, including other tuples.

`````python
my_tuple = (1, 2, 3, 4)
print(my_tuple)
`````

`````python
mixed_tuple = (1, "hello", 3.14, True)
print(mixed_tuple)
`````

`````python
my_tuple = (1, 2, 3, 4)
print(my_tuple[0])
print(my_tuple[1])
print(my_tuple[-1])
`````

`````python
# Slicing

my_tuple = (1, 2, 3, 4, 5)
print(my_tuple[1:4])
print(my_tuple[:3])
print(my_tuple[3:])
`````

`````python
#Concatenating Tuples:

tuple1 = (1, 2, 3)
tuple2 = (4, 5, 6)
combined_tuple = tuple1 + tuple2
print(combined_tuple)
`````

`````python
my_tuple = (1, 2, 3)
repeated_tuple = my_tuple * 3
print(repeated_tuple)
`````

`````python
# Unpacking

my_tuple = (1, 2, 3)
a, b, c = my_tuple
print(a)
print(b)
print(c)
`````

`````python
# Count method

my_tuple = (1, 2, 3, 1, 1)
count_1 = my_tuple.count(1)

print(count_1)
`````

### Dictionaries

In Python, a dictionary is a built-in data type that allows you to store data in key-value pairs. Dictionaries are unordered collections, meaning that the items do not have a specific order. Each key in a dictionary is unique and maps to a specific value.

Key Characteristics of Dictionaries:


*   Unordered: Dictionaries do not maintain the order of items.
*   Key-Value Pairs: Each item is a pair consisting of a key and a value.
*   Mutable: You can change, add, or remove key-value pairs.
*   Keys are Unique: Each key must be unique within a dictionary.

`````python
d = {'key1':'item1','key2':'item2'}
`````

`````python
person = {
    "name": "John",
    "age": 25,
    "city": "London"
}
print(person)
`````

`````python
person = {
    "name": "John",
    "age": 25,
    "city": "London"
}
print(person["name"])
print(person["age"])
`````

`````python
person = {
    "name": "John",
    "age": 25,
    "city": "London"
}
person["age"] = 26
print(person)
`````

`````python
person = {
    "name": "John",
    "age": 25,
    "city": "London"
}
del person["city"]
print(person)
`````

`````python
person = {
    "name": "John",
    "age": 25
}
print(person.get("name"))
print(person.get("city", "Not Found"))
`````

`````python
person = {
    "name": "John",
    "age": 25
}
print("name" in person)
print("city" in person)
`````

`````python
person = {
    "name": "John",
    "age": 25,
    "city": "London"
}

for key, value in person.items():
    print(f"Key: {key}, Value: {value}")
`````
